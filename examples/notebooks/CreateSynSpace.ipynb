# Create SynSpace based on a molecule

```python
import synspace
```

```python
smi: str = "CCCO"
mols, props = synspace.chemical_space(smi, num_samples=3000, filter=False)
```

```python
from rdkit.Chem import Draw

print(len(mols))
Draw.MolsToGridImage(mols, molsPerRow=4, subImgSize=(200, 200))
```
